```python
#7.1
import tkinter as tk
from tkinter import messagebox

window = tk.Tk()
window.title("Income Tax Calculator")
window.geometry("400x500")

label1 = tk.Label(window, text="Enter Annual Income:")
label1.pack(pady=(10, 5))

annualIncome = tk.Entry(window)
annualIncome.pack()

regime_label = tk.Label(window, text="Select Tax Regime:")
regime_label.pack(pady=(15, 5))

regime = tk.StringVar()
regime.set("New Regime")
#radio buttons
old_regime = tk.Radiobutton(
    window,
    text="Old Regime",
    variable=regime,
    value="Old Regime"
)
old_regime.pack()

new_regime = tk.Radiobutton(
    window,
    text="New Regime",
    variable=regime,
    value="New Regime"
)
new_regime.pack()

def calculate_tax():
    try:
        income = int(annualIncome.get())
        if income < 0:
            messagebox.showwarning(
                "Invalid Input",
                "Income cannot be negative"
            )
            return
        selected_regime = regime.get()
        if selected_regime == "New Regime":
            if income <= 400000:
                taxAmount = 0
            elif income <= 800000:
                taxableAmount = income - 400000
                taxAmount = taxableAmount * 0.05
            elif income <= 1200000:
                taxableAmount = income - 400000
                taxAmount = (taxableAmount - 400000) * 0.10 + 400000 * 0.05
            elif income <= 1600000:
                taxableAmount = income - 400000
                taxAmount = (
                    400000 * 0.05
                    + 400000 * 0.10
                    + (taxableAmount - 800000) * 0.15
                )
            else:
                taxableAmount = income - 400000
                taxAmount = (
                    400000 * 0.05
                    + 400000 * 0.10
                    + 400000 * 0.15
                    + (taxableAmount - 1200000) * 0.20
                )
        else:
            if income <= 250000:
                taxAmount = 0
            elif income <= 500000:
                taxableAmount = income - 250000
                taxAmount = taxableAmount * 0.05
            elif income <= 1000000:
                taxableAmount = income - 250000
                taxAmount = (
                    250000 * 0.05
                    + (taxableAmount - 250000) * 0.20
                )
            else:
                taxableAmount = income - 250000
                taxAmount = (
                    250000 * 0.05
                    + 500000 * 0.20
                    + (taxableAmount - 750000) * 0.30
                )

        result_label.config(
            text="Result:\nIncome Tax: ₹" + str(round(taxAmount, 2))
            +"\n Annual Income: "+annualIncome.get()
            +"\nRegime selected : "+selected_regime
        )
    except ValueError:
        messagebox.showwarning(
            "Invalid Input",
            "Please enter a valid annual income"
        )
calculate_button = tk.Button(
    window,
    text="Calculate Tax",
    command=calculate_tax
)
calculate_button.pack(pady=20)
result_label = tk.Label(
    window,
    text="",
    font=("Arial", 12)
)
result_label.pack(pady=10)
def reset():
    annualIncome.delete(0, tk.END)
    regime.set("New Regime")
    result_label.config(text="Income Tax:")
reset_button = tk.Button(
    window,
    text="Reset",
    command=reset
)
reset_button.pack(pady=10)

window.mainloop()
```

```python
#2
import tkinter as tk
from tkinter import messagebox

window = tk.Tk()
window.title("Simple Calculator")
window.geometry("350x450")

display = tk.Entry(
    window,
    font=("Arial", 20),
    justify="right"
)
display.grid(row=0, column=0, columnspan=4, padx=10, pady=20)

def button_click(value):
    display.insert(tk.END, value)
def clear():
    display.delete(0, tk.END)
def calculate():
    expression = display.get()
    try:
        result = eval(expression)#for evaluating math expression string
        display.delete(0, tk.END)
        display.insert(0, result)

    except ZeroDivisionError:
        messagebox.showerror(
            "Error",
            "Cannot divide by zero"
        )
        clear()
    except:
        messagebox.showerror(
            "Error",
            "Invalid operation"
        )
        clear()

buttons = [
    ("7", 1, 0),
    ("8", 1, 1),
    ("9", 1, 2),
    ("/", 1, 3),

    ("4", 2, 0),
    ("5", 2, 1),
    ("6", 2, 2),
    ("*", 2, 3),

    ("1", 3, 0),
    ("2", 3, 1),
    ("3", 3, 2),
    ("-", 3, 3),

    ("0", 4, 0),
    (".", 4, 1),
    ("+", 4, 2),
    ("=", 4, 3)
]

for text, row, column in buttons:
    if text == "=":
        button = tk.Button(
            window,
            text=text,
            width=7,
            height=2,
            command=calculate
        )
    else:
        button = tk.Button(
            window,
            text=text,
            width=7,
            height=2,
            command=lambda value=text: button_click(value)
            #when this button is clciked ,send its text to button_click()
        )
    button.grid(
        row=row,
        column=column,
        padx=5,
        pady=5
    )
clear_button = tk.Button(
    window,
    text="Clear",
    width=31,
    height=2,
    command=clear
)
clear_button.grid(
    row=5,
    column=0,
    columnspan=4,
    padx=5,
    pady=10
)
window.mainloop()
```
