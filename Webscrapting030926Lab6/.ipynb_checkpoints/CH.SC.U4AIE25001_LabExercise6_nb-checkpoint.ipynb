```python
#CH.SC.U4AIE25001
# 1.Add one more column to the existing quotes dataset to store the tags associated with each quote.
import requests
from bs4 import BeautifulSoup
import pandas as pd


url = "https://quotes.toscrape.com/"
response = requests.get(url)


soup = BeautifulSoup(response.text, "html.parser")
quote_blocks = soup.find_all("div", class_="quote")

completeData = []
for block in quote_blocks:
    text = block.find("span", class_="text").text
    author = block.find("small", class_="author").text
    tags = []
    for tag in block.find_all("a", class_="tag"):
        tags.append(tag.text)
    # print(text," - ",author," - ",tags)
    completeData.append([text,author,tags])

df = pd.DataFrame(completeData, columns=["Quote","Author","Tags"])
df

```

```python
#2.Filter and display only the quotes where the tag is "love".
import requests
from bs4 import BeautifulSoup
import pandas as pd


url = "https://quotes.toscrape.com/"
response = requests.get(url)


soup = BeautifulSoup(response.text, "html.parser")
quote_blocks = soup.find_all("div", class_="quote")

completeData = []
for block in quote_blocks:
    text = block.find("span", class_="text").text
    author = block.find("small", class_="author").text
    tags = []
    for tag in block.find_all("a", class_="tag"):
        tags.append(tag.text)
    # print(text," - ",author," - ",tags)
    completeData.append([text,author,tags])

df = pd.DataFrame(completeData, columns=["Quote","Author","Tags"])
# rows_withlovetag = df[df["Tags"].str.contains("love")]
rows_with_love_tag = df[df["Tags"].apply(lambda tags: "love" in tags)]
rows_with_love_tag
```

```python
Assignment 3: BookMyShow movie data
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://in.bookmyshow.com/explore/home/chennai"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/114.0.0.0 Safari/537.36"
}

response = requests.get(url, headers=headers)
print("Status Code:", response.status_code)
soup = BeautifulSoup(response.text, "html.parser")
movies_data = []
movie_links = soup.find_all("a", class_="cnWwcT")

for link in movie_links:
    movie_url = link.get("href")
    if movie_url and "/movies/" in movie_url:

        # Movie Name
        name_element = link.find("div", class_="fpkEDc")
        if name_element:
            movie_name = name_element.text.strip()
        else:
            movie_name = "N/A"
        # Genre
        genre_element = link.find("div", class_="engTlI")
        if genre_element:
            genre = genre_element.text.strip()
        else:
            genre = "N/A"
        # Complete Movie URL
        if movie_url.startswith("/"):
            movie_url = "https://in.bookmyshow.com" + movie_url
        # Movie ID
        movie_id = movie_url.split("/")[-1]
        if movie_name != "N/A":
            movies_data.append([
                movie_name,
                genre,
                movie_id,
                movie_url
            ])


# Create DataFrame
df_movies = pd.DataFrame(
    movies_data,
    columns=[
        "Movie Name",
        "Genre",
        "Movie ID",
        "Movie URL"
    ]
)

# Save as CSV
df_movies.to_csv(
    "bms_chennai_movies.csv",
    index=False
)


print(df_movies.head())
print("\nTotal movies scraped:", len(df_movies))
print("\nCSV file saved as bms_chennai_movies.csv")
```
