## 2. Variables & Data Types
Python is **dynamically typed** — you never declare a type, Python figures it out automatically.

```python
age = 20          # int
gpa = 8.75        # float
name = "Alpha"    # str
passed = True     # bool

print(age, type(age))
print(gpa, type(gpa))
print(name, type(name))
print(passed, type(passed))
```

## 3. Decision Making — if / elif / else

```python
marks = 78

if marks >= 90:
    print("Grade: O (Outstanding)")
elif marks >= 80:
    print("Grade: A+ (Excellent)")
elif marks >= 70:
    print("Grade: A (Very Good)")
elif marks >= 60:
    print("Grade: B+ (Good)")
else:
    print("Grade: B (Average)")
```

*Try changing `marks` above and re-running the cell to see different grades.*

## 4. for Loop — Repeating for Every Item in a Sequence

```python
fruits = ["apple", "banana", "mango"]
for fruit in fruits:
    print(fruit)

print("Loop finished!")
```

## 5. range() — Generating Number Sequences

```python
print("range(5):")
for i in range(5):
    print(i, end=" ")
print()

print("range(2, 7):")
for i in range(2, 7):
    print(i, end=" ")
print()

print("range(0, 10, 2):")
for i in range(0, 10, 2):
    print(i, end=" ")
print()

print("Countdown range(10, 0, -1):")
for i in range(10, 0, -1):
    print(i, end=" ")
print()
```

## 6. while Loop — Repeat as Long as Condition is True

```python
count = 1
while count <= 5:
    print("Count:", count)
    count = count + 1   # must change, or infinite loop!

print("Loop done!")
```

### Interactive demo — password guesser (uses `input()`, run this live)

```python
password = "python"
guess = input("Enter password: ")
while guess != password:
    print("Wrong! Try again.")
    guess = input("Enter password: ")
print("Correct! Access granted.")
```

## 7. break · continue · pass — Controlling Loop Flow

```python
# break: exit the loop immediately
for i in range(1, 10):
    if i == 5:
        break
    print(i, end=" ")
print("\nStopped!")
```

```python
# continue: skip this iteration, keep looping
for i in range(1, 8):
    if i % 2 == 0:
        continue
    print(i, end=" ")
```

```python
# pass: do nothing, placeholder for later
for i in range(5):
    pass   # will add code later
print("Done")
```

## 8. Nested Loops — A Loop Inside Another Loop

```python
# Multiplication table
for i in range(1, 4):
    for j in range(1, 4):
        result = i * j
        print(i, "x", j, "=", result, end="  ")
    print()
```

### Classic interview problem — FizzBuzz

```python
for n in range(1, 16):
    if n % 15 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
```
