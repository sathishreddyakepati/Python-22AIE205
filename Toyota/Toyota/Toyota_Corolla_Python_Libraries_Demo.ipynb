# 9. DataFrame Column Operations

The following examples use common Toyota Corolla columns such as `Price`, `KM`, and `HP`.

The helper below finds the actual column name without changing your dataset.

```python
def find_column(df, candidates):
    lookup = {str(c).strip().lower(): c for c in df.columns}
    for candidate in candidates:
        if candidate.lower() in lookup:
            return lookup[candidate.lower()]
    return None

price_col = find_column(df, ["Price", "price"])
km_col = find_column(df, ["KM", "km", "Kilometers", "Kilometres"])
hp_col = find_column(df, ["HP", "hp", "Horsepower"])
fuel_col = find_column(df, ["Fuel_Type", "Fuel Type", "FuelType", "fuel_type"])
age_col = find_column(df, ["Age_08_04", "Age", "age"])

print("Price column:", price_col)
print("KM column:", km_col)
print("HP column:", hp_col)
print("Fuel Type column:", fuel_col)
print("Age column:", age_col)
```

## 10. Select One Column

If the dataset contains `Price`, we can select that column as follows.

```python
if price_col:
    display(df[[price_col]].head())
else:
    print("Price column was not found. Check df.columns.")
```

## 11. Select Multiple Columns

```python
available = [c for c in [price_col, km_col, hp_col] if c is not None]

if available:
    display(df[available].head())
else:
    print("None of the expected numerical columns were found.")
```

# 12. Filtering Data

### Example: Cars with price greater than 10,000

```python
if price_col:
    expensive_cars = df[df[price_col] > 10000]
    display(expensive_cars.head(10))
else:
    print("Price column was not found.")
```

### Example: Cars with fewer than 50,000 kilometres

```python
if km_col:
    low_km_cars = df[df[km_col] < 50000]
    display(low_km_cars.head(10))
else:
    print("KM column was not found.")
```

### Combine two conditions

Find cars with **price greater than 10,000 AND KM less than 50,000**.

```python
if price_col and km_col:
    result = df[
        (df[price_col] > 10000) &
        (df[km_col] < 50000)
    ]
    display(result.head(10))
else:
    print("Required columns were not found.")
```

# 13. Sorting

### Sort by Price – Lowest to Highest

```python
if price_col:
    display(df.sort_values(price_col).head(10))
```

### Sort by Price – Highest to Lowest

```python
if price_col:
    display(df.sort_values(price_col, ascending=False).head(10))
```

# 14. Basic Statistical Operations

```python
if price_col:
    print("Average Price:", df[price_col].mean())
    print("Maximum Price:", df[price_col].max())
    print("Minimum Price:", df[price_col].min())

if km_col:
    print("Average KM:", df[km_col].mean())

if hp_col:
    print("Average HP:", df[hp_col].mean())
```

# 15. Value Counts

`value_counts()` counts how many records belong to each category.

For the Corolla dataset, this is useful for `Fuel_Type`.

```python
if fuel_col:
    print(df[fuel_col].value_counts())
else:
    print("Fuel type column was not found.")
```

# 16. GroupBy Operation

`groupby()` allows us to divide data into groups and calculate statistics for each group.

Example: average price for each fuel type.

```python
if fuel_col and price_col:
    group_price = df.groupby(fuel_col)[price_col].mean().sort_values(ascending=False)
    display(group_price)
else:
    print("Fuel type or Price column was not found.")
```

### Average kilometres by fuel type

```python
if fuel_col and km_col:
    group_km = df.groupby(fuel_col)[km_col].mean().sort_values(ascending=False)
    display(group_km)
else:
    print("Fuel type or KM column was not found.")
```

# 17. Data Visualization using Matplotlib

We will now convert the DataFrame information into visual form.

## Histogram – Price Distribution

```python
if price_col:
    plt.figure(figsize=(8, 5))
    plt.hist(df[price_col].dropna(), bins=20)
    plt.xlabel("Price")
    plt.ylabel("Number of Cars")
    plt.title("Distribution of Toyota Corolla Prices")
    plt.show()
else:
    print("Price column was not found.")
```

## Scatter Plot – Price vs Kilometres

A scatter plot helps us examine the relationship between two numerical variables.

```python
if price_col and km_col:
    plt.figure(figsize=(8, 5))
    plt.scatter(df[km_col], df[price_col], alpha=0.6)
    plt.xlabel("Kilometres")
    plt.ylabel("Price")
    plt.title("Toyota Corolla Price vs Kilometres")
    plt.show()
else:
    print("Price or KM column was not found.")
```

## Bar Chart – Cars by Fuel Type

```python
if fuel_col:
    fuel_counts = df[fuel_col].value_counts()

    plt.figure(figsize=(7, 5))
    plt.bar(fuel_counts.index.astype(str), fuel_counts.values)
    plt.xlabel("Fuel Type")
    plt.ylabel("Number of Cars")
    plt.title("Toyota Corolla Cars by Fuel Type")
    plt.show()
else:
    print("Fuel type column was not found.")
```

# 18. Data Visualization using Seaborn

Seaborn provides a high-level interface for statistical graphics.

## Seaborn Histogram

```python
if price_col:
    plt.figure(figsize=(8, 5))
    sns.histplot(data=df, x=price_col, bins=20)
    plt.title("Distribution of Toyota Corolla Prices")
    plt.show()
else:
    print("Price column was not found.")
```

## Seaborn Scatter Plot

```python
if price_col and km_col:
    plt.figure(figsize=(8, 5))
    sns.scatterplot(data=df, x=km_col, y=price_col)
    plt.title("Toyota Corolla Price vs Kilometres")
    plt.show()
else:
    print("Price or KM column was not found.")
```

## Seaborn Box Plot – Price by Fuel Type

A box plot helps compare the distribution of a numerical variable across categories.

```python
if fuel_col and price_col:
    plt.figure(figsize=(8, 5))
    sns.boxplot(data=df, x=fuel_col, y=price_col)
    plt.title("Price Distribution by Fuel Type")
    plt.show()
else:
    print("Fuel type or Price column was not found.")
```

# 19. Correlation Heatmap

Correlation helps us examine relationships between numerical variables.

We select only numerical columns before calculating the correlation matrix.

```python
numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(correlation, annot=True, fmt=".2f")
plt.title("Toyota Corolla Numerical Feature Correlation")
plt.show()
```
